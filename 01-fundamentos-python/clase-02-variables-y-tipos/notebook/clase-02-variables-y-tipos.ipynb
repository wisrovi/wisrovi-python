## 1. 💡 Fundamentación Teórica y Modelo Mental

En Python, las variables almacenan referencias (punteros) a objetos en el Heap de memoria.
Siguiendo el **Aprendizaje en Espiral**, integramos las variables con **funciones (`def`)** para crear transformaciones robustas y reutilizables.

> **🌟 Metáfora Didáctica:** Una variable es una etiqueta adhesiva pegada a una caja; varias etiquetas pueden apuntar a la misma caja. Una función es *«La Licuadora»*, que recibe parámetros tipados (PEP 484), procesa los datos y devuelve un nuevo objeto en el Heap.

**Principios Clave:**
1. **Tipado Dinámico y Fuerte:** No realiza conversiones implícitas incompatibles.
2. **Type Hints (PEP 484):** Documenta contratos de entrada (`x: float`) y salida (`-> float`).
3. **Inmutabilidad:** Modificar `int`, `float`, `str` o `bool` genera un nuevo objeto en memoria.

**⚡ Regla de Oro:** Convierte tipos explícitamente usando `int()` o `float()` antes de operar con entradas de usuario.


## 2. 💻 Implementación de Código con Funciones y Type Hints

Ejecuta la siguiente celda para observar el cálculo financiero encapsulado en una función tipada:


```python
# CLASE 02 - Demostración de Funciones con Type Hints y Casting
def calcular_resumen_pedido(
    producto: str, 
    precio_unitario_str: str, 
    cantidad: int, 
    tasa_iva: float = 0.21
) -> str:
    """Procesa una orden con casting seguro y genera un recibo formateado."""
    precio_unitario: float = float(precio_unitario_str)
    subtotal: float = precio_unitario * cantidad
    monto_iva: float = subtotal * tasa_iva
    total_final: float = subtotal + monto_iva
    
    return (
        f"--- RESUMEN DE COMPRA ---\n"
        f"Producto:     {producto:<20}\n"
        f"Cantidad:     {cantidad:>5}\n"
        f"Subtotal:     ${subtotal:>8.2f}\n"
        f"IVA ({tasa_iva*100:.0f}%):    ${monto_iva:>8.2f}\n"
        f"Total Final:  ${total_final:>8.2f}"
    )

factura = calcular_resumen_pedido("Teclado Mecánico RGB", "79.99", 2)
print(factura)
```

## 4. 🏋️ Desafío Práctico: Calculadora de Propinas Modular

> **Reto:** Implementa las 3 funciones tipadas requeridas:
> 1. `calcular_propina(total_cuenta: float, porcentaje: float) -> float`
> 2. `calcular_total_por_persona(total_cuenta: float, porcentaje: float, num_personas: int) -> float`
> 3. `formatear_factura(total_cuenta: float, propina: float, total_por_persona: float) -> str`

¡Escribe y prueba tu solución en la siguiente celda interactiva!


```python
# Escribe tu solución aquí:
def calcular_propina(total_cuenta: float, porcentaje: float) -> float:
    propina = total_cuenta * (porcentaje / 100.0)
    return round(propina, 2)

def calcular_total_por_persona(total_cuenta: float, porcentaje: float, num_personas: int) -> float:
    if num_personas <= 0:
        raise ValueError("El número de personas debe ser mayor a 0.")
    monto_propina = calcular_propina(total_cuenta, porcentaje)
    return round((total_cuenta + monto_propina) / num_personas, 2)

def formatear_factura(total_cuenta: float, propina: float, total_por_persona: float) -> str:
    return (
        f"--- RESUMEN DE CUENTA ---\n"
        f"Subtotal:         ${total_cuenta:.2f}\n"
        f"Propina:          ${propina:.2f}\n"
        f"Total a Pagar:    ${total_cuenta + propina:.2f}\n"
        f"Cuota Individual: ${total_por_persona:.2f}"
    )

print(formatear_factura(100.0, calcular_propina(100.0, 15.0), calcular_total_por_persona(100.0, 15.0, 2)))
```
